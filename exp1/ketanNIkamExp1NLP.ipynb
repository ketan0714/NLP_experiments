## 1. Import Libraries and Load spaCy

```python
import spacy
import nltk
import pandas as pd
from nltk.stem import PorterStemmer, SnowballStemmer

print("spaCy version:", spacy.__version__)
print("NLTK version:", nltk.__version__)

nlp = spacy.load("en_core_web_sm")
print("spaCy English model loaded successfully.")
```

## 2. Step 1 – Enter the Given Text

```python
text =  """
The children are playing in the gardens while their parents watch them.
A researcher studies language patterns and analyzes different documents.
The students were writing, reading, and discussing their experiments.
The machines process data, processed old records, and are processing new files.
Good preprocessing techniques help computers understand words and meanings.
""".strip()

print(text)
print("\nCharacter count:", len(text))
```

## Step 2 – Create the spaCy Document


```python
doc = nlp(text)

print("Number of spaCy tokens:", len(doc))

print("First 30 spaCy tokens: -   ")

print([token.text for token in doc[:30]])
```

## 3.Perform Different Tokenization Methods

### 3.1 Whitespace Tokenization
Whitespace tokenization simply splits the text wherever whitespace occurs. Therefore, punctuation may remain attached to words such as language.


```python
whitespace_tokens = text.split()

print("Whitespace tokens:")
print(whitespace_tokens )

print("\nCount:", len(whitespace_tokens))
```

### 3.2 spaCy Word Tokenization
spaCy uses linguistic tokenization rules and separates punctuation from words.

```python
spacy_tokens = [token.text for token in doc]

print("spaCy tokens:")
print(spacy_tokens)

print("\nCount:", len(spacy_tokens))
```

### 3.2 spaCy Tokens Without Punctuation and Spaces
This filtered list is used for stemming, lemmatization, and vocabulary analysis.

```python
filtered_doc_tokens = [
    token for token in doc
    if not token.is_space and not token.is_punct
]

clean_tokens = [token.text for token in filtered_doc_tokens]

print("Before filtering:", len(spacy_tokens))
print("After filtering: ", len(clean_tokens))
print("Removed:         ", len(spacy_tokens) - len(clean_tokens))

print("\nFiltered tokens:")
print(clean_tokens)
```

## 4. Tokenization Comparison

```python
tokenization_summary = pd.DataFrame({
    "Method": [
        "Whitespace tokenization",
        "spaCy tokenization",
        "spaCy filtered tokenization"
    ],
    "Token Count": [
        len(whitespace_tokens),
        len(spacy_tokens),
        len(clean_tokens)
    ]
})

display(tokenization_summary)
```

## 5. Stemming
Stemming applies heuristic rules and may produce a string that is not a valid English word. The lab manual specifically asks for both PorterStemmer and SnowballStemmer.

### 5.1 PorterStemmer


```python
porter = PorterStemmer()

original_words = [
    token.text.lower()
    for token in filtered_doc_tokens
    if token.is_alpha
]

porter_stems = [porter.stem(word) for word in original_words]

print("Porter stems:")
print(porter_stems)
```

### 5.2 SnowballStemmer

```python
snowball = SnowballStemmer("english")

original_words = [
    token.text.lower()
    for token in filtered_doc_tokens
    if token.is_alpha
]

snowball_stems = [snowball.stem(word) for word in original_words]


print("\nSnowball stems:")
print(snowball_stems)
```

### 5.3 Porter vs Snowball – Different Outputs
This cell identifies words for which the two stemmers produce different stems.

```python
stemmer_differences = pd.DataFrame([
    {
        "Word": word,
        "Porter": porter.stem(word),
        "Snowball": snowball.stem(word)
    }
    for word in original_words
    if porter.stem(word) != snowball.stem(word)
])

if stemmer_differences.empty:
    print("No Porter/Snowball differences found for this text.")
else:
    display(stemmer_differences.drop_duplicates().reset_index(drop=True))
```

## 6.  Lemmatization Using spaCy
Lemmatization maps words to their dictionary/base forms using linguistic information. spaCy's model also provides POS information, which helps explain why lemmatization is generally more interpretable than heuristic stemming.

```python
lemmas = [
    token.lemma_.lower()
    for token in filtered_doc_tokens
    if token.is_alpha
]

print("spaCy lemmas:")
print(lemmas)
```

## 7. POS, Stem and Lemma Inspection
The lab manual recommends inspecting POS and lemmas together to understand the linguistic difference between stemming and lemmatization.

```python
word_analysis = pd.DataFrame({
    "Original": [token.text for token in filtered_doc_tokens if token.is_alpha],
    "POS": [token.pos_ for token in filtered_doc_tokens if token.is_alpha],
    "Porter Stem": [porter.stem(token.text.lower())
                    for token in filtered_doc_tokens if token.is_alpha],
    "Snowball Stem": [snowball.stem(token.text.lower())
                      for token in filtered_doc_tokens if token.is_alpha],
    "spaCy Lemma": [token.lemma_.lower()
                    for token in filtered_doc_tokens if token.is_alpha]
})

display(word_analysis)
```

## 8. Comparison Table for Selected Words
These examples follow the lab manual's expected observations: **studying, studied, processing**, and **systems**.

```python
selected_words = [
    "writing",
    "processed",
    "discussing",
    "computers"
]

selected_rows = []

for word in selected_words:
    matches = [
        token for token in filtered_doc_tokens
        if token.is_alpha and token.text.lower() == word
    ]

    if matches:
        token = matches[0]
        selected_rows.append({
            "Original": token.text,
            "POS": token.pos_,
            "Porter Stem": porter.stem(word),
            "Snowball Stem": snowball.stem(word),
            "spaCy Lemma": token.lemma_.lower()
        })

selected_comparison = pd.DataFrame(selected_rows)
display(selected_comparison)
```

## 9. Compare Word Variations
A lower number of unique normalized forms indicates stronger vocabulary reduction, but it does not automatically mean better preprocessing. The normalized forms must also remain meaningful.

For stemming and lemmatization, the reduction percentage is calculated using the formula from the lab manual:

Reduction % = (1 - normalized unique forms / original unique forms) × 100

```python
def reduction_percentage(original, normalized):
    original_unique = len(set(original))
    normalized_unique = len(set(normalized))

    if original_unique == 0:
        return 0.0

    return (1 - normalized_unique / original_unique) * 100

porter_reduction = reduction_percentage(original_words, porter_stems)
snowball_reduction = reduction_percentage(original_words, snowball_stems)
lemma_reduction = reduction_percentage(original_words, lemmas)

variation_metrics = pd.DataFrame({
    "Method": [
        "Original words",
        "Porter stemming",
        "Snowball stemming",
        "spaCy lemmatization"
    ],
    "Total Words": [
        len(original_words),
        len(porter_stems),
        len(snowball_stems),
        len(lemmas)
    ],
    "Unique Forms": [
        len(set(original_words)),
        len(set(porter_stems)),
        len(set(snowball_stems)),
        len(set(lemmas))
    ],
    "Reduction %": [
        0.0,
        round(porter_reduction, 2),
        round(snowball_reduction, 2),
        round(lemma_reduction, 2)
    ]
})

display(variation_metrics)
```

## 10. Tokenization/Stemming/Lemmatization Result Table
The lab manual's result table asks for total tokens/words, unique forms, reduction percentage, and semantic coherence.

For tokenization, reduction percentage is not a meaningful normalization measure because tokenization changes token boundaries rather than normalizing word forms. Therefore, it is reported as N/A instead of inventing a misleading value.

```python
result_table = pd.DataFrame({
    "Method": [
        "Whitespace tokenization",
        "spaCy tokenization",
        "Porter stemming",
        "Snowball stemming",
        "spaCy lemmatization"
    ],
    "Total Tokens/Words": [
        len(whitespace_tokens),
        len(spacy_tokens),
        len(porter_stems),
        len(snowball_stems),
        len(lemmas)
    ],
    "Unique Forms": [
        len(set(whitespace_tokens)),
        len(set(spacy_tokens)),
        len(set(porter_stems)),
        len(set(snowball_stems)),
        len(set(lemmas))
    ],
    "Reduction %": [
        "N/A",
        "N/A",
        round(porter_reduction, 2),
        round(snowball_reduction, 2),
        round(lemma_reduction, 2)
    ],
    "Semantic Coherence": [
        "Not applicable",
        "Not applicable",
        "Lower – stems may be truncated/non-words",
        "Lower – stems may be truncated/non-words",
        "Higher – generally interpretable base forms"
    ]
})

display(result_table)
```

## 11. Qualitative Semantic Coherence
The lab manual asks us to evaluate semantic coherence qualitatively. We check whether:

normalized forms are valid/interpretable English words,
related forms such as study, studying, and studied are reasonably normalized,
stems look unnatural or truncated,
lemmas preserve recognizable base forms.
A vocabulary reduction percentage alone is not sufficient to claim semantic superiority.

```python
# Representative examples for qualitative observation

examples = ["writing", "processing", "discussing", "computers"]

observation_rows = []

for word in examples:
    token = next(
        (t for t in filtered_doc_tokens
         if t.is_alpha and t.text.lower() == word),
        None
    )

    if token is not None:
        observation_rows.append({
            "Original": word,
            "Porter Stem": porter.stem(word),
            "Snowball Stem": snowball.stem(word),
            "spaCy Lemma": token.lemma_.lower(),
            "Observation": (
                "Lemma is a recognizable base word; stems may be truncated."
                if word in {"studying", "studied"}
                else
                "Normalization produces an interpretable form."
                if word == "processing"
                else
                "Plural normalization is clear."
            )
        })

observation_df = pd.DataFrame(observation_rows)
display(observation_df)
```

The lab manual uses this short sentence to demonstrate the difference between stemming and lemmatization.

```python
sample = "The children are playing in the gardens while their parents watch them."
sample_doc = nlp(sample)

sample_words = [token for token in sample_doc if token.is_alpha]

sample_porter = [porter.stem(token.text.lower()) for token in sample_words]
sample_snowball = [snowball.stem(token.text.lower()) for token in sample_words]
sample_lemmas = [token.lemma_.lower() for token in sample_words]

print("Original:")
print(sample)

print("\nPorter stems:")
print(sample_porter)

print("\nSnowball stems:")
print(sample_snowball)

print("\nspaCy lemmas:")
print(sample_lemmas)
```
